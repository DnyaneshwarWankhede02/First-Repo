# **Reading from csv file**

```python
df = spark.read.format("csv").option("header", "true").option("inferSchema","true").load("/Volumes/workspace/bronze/source_system/sales_details.csv")

df.display()
```

# write to bronze layer

```python
df.write.mode("overwrite").saveAsTable("workspace.bronze.crm_sales_details")
```

```python
%sql
select * from workspace.bronze.crm_sales_details
```
